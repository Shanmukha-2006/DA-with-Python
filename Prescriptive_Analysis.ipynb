# Prescriptive Analysis

# User Data

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "User": ["Ravi", "Sita", "Ram", "Anu"],
    "Purchase_Amount": [5000, 1500, 7000, 2000],
    "Visits": [10, 3, 15, 4],
    "Rating": [5, 3, 4, 2]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)


recommendation = []

for i in range(len(df)):

    if df["Purchase_Amount"][i] > 4000 and df["Rating"][i] >= 4:
        recommendation.append("Recommend Premium Products")

    elif df["Visits"][i] < 5:
        recommendation.append("Send Discount Offer")

    else:
        recommendation.append("Recommend Regular Products")

df["Recommendation"] = recommendation

print("\nPrescriptive Analysis Result:")
print(df)
```

```python
#line chart
plt.plot(df["User"], df["Purchase_Amount"])
plt.xlabel("User")
plt.ylabel("Purchase Amount")
plt.title("Purchase Amount by User")
plt.show()
```

```python
plt.bar(df["User"], df["Visits"])
plt.xlabel("User")
plt.ylabel("Visits")
plt.title("Visits by User")
plt.show()
```

```python
plt.pie(df["Rating"], labels=df["User"], autopct="%1.1f%%")
plt.title("Rating Distribution")
plt.show()
```

```python
plt.scatter(df["Purchase_Amount"], df["Visits"])
plt.xlabel("Purchase Amount")
plt.ylabel("Visits")
plt.title("Purchase Amount vs Visits")
plt.show()
```

```python
plt.hist(df["Purchase_Amount"], bins=10)
plt.xlabel("Purchase Amount")
plt.ylabel("Frequency")
plt.title("Purchase Amount Distribution")
plt.show()
```

```python
plt.boxplot(df["Purchase_Amount"])
plt.ylabel("Purchase Amount")
plt.title("Purchase Amount Boxplot")
plt.show()
```

# From Kaggle Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("/content/patients.csv")
print(df)

print(df.head())

df["Recommendation"] = df.apply(
    lambda row:
    "Senior Care Advice" if row["age"] > 60 else
    "Improve Patient Experience" if row["satisfaction"] < 70 else
    "Standard Care Plan",
    axis=1
)

# Display recommendations
print("\nPatient Recommendations:\n")
print(df[["age", "satisfaction", "Recommendation"]])

# Count recommendations
recommendation_count = df["Recommendation"].value_counts()

print("\nRecommendation Counts:\n")
print(recommendation_count)
```

```python
recommendation_counts = df['Recommendation'].value_counts()
plt.pie(recommendation_counts.values, labels=recommendation_counts.index, autopct="%1.2f%%")
plt.title("Recommendation Distribution")
plt.show()
```

```python
plt.hist(df["satisfaction"], bins=10)
plt.xlabel("Satisfaction")
plt.ylabel("Frequency")
plt.title("Satisfaction Distribution")
plt.show()
```

```python
plt.scatter(df["age"], df["satisfaction"])
plt.xlabel("Age")
plt.ylabel("Satisfaction")
plt.title("Age vs Satisfaction")
plt.show()
```

```python
plt.boxplot(df["satisfaction"])
plt.ylabel("Satisfaction")
plt.title("Satisfaction Boxplot")
plt.show()
```

```python
plt.bar(df["age"], df["satisfaction"])
plt.xlabel("Age")
plt.ylabel("Satisfaction")
plt.title("Satisfaction by Age")
plt.show()
```

```python
plt.plot(df["service"], df["age"])
plt.xlabel("service")
plt.ylabel("Age")
plt.title("Age by service")
plt.show()
```
