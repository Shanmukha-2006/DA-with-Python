# Prescriptive Analysis

# User Data

```python
import pandas as pd
import matplotlib.pyplot as plt
data=pd.DataFrame({
    "patient_name":["sunil","vardhan","thiru","vishnu","shanmukha"],
    "age":[33,12,27,25,21],
    "condition":["very serious","safe","serious","serious","safe"],
    "oxygen_level":[100,70,56,80,96],
    "available_beds":20,
})
print(data)
score = []
for i in range(len(data)):
    s = 0
    if data.loc[i, "age"] > 60:
        s += 2
    if data.loc[i, "oxygen_level"] < 85:
        s += 3
    score.append(s)
data["PriorityScore"] = score
bed = []
for i in data["PriorityScore"]:
    if i >= 3:
       bed.append("Allocated")
    else:
        bed.append("Not Allocated")
data["Bed"] = bed
prescription = []
for i in data["PriorityScore"]:
    if i >= 5:
       prescription.append("ICU")
    elif i >= 3:
        prescription.append("Oxygen Support")
    else:
        prescription.append("Normal Checkup")
data["Prescription"] = prescription
print(data)
```

```python
#line chart
plt.plot(data["patient_name"],data["age"])
plt.xlabel("patient_name")
plt.ylabel("age")
plt.title("patient_name vs age")
plt.show()
```

```python
plt.bar(data["patient_name"],data["oxygen_level"])
plt.xlabel("patient_name")
plt.ylabel("oxygen_level")
plt.title("patient_name vs oxygen_level")
plt.show()
```

```python
plt.scatter(data["patient_name"],data["oxygen_level"])
plt.xlabel("patient_name")
plt.ylabel("oxygen_level")
plt.title("patient_name vs oxygen_level")
plt.show()
```

```python
plt.hist(data["oxygen_level"],bins=10)
plt.xlabel("oxygen_level")
plt.ylabel("frequency")
plt.title("oxygen_level distribution")
plt.show()
```

```python
plt.pie(data["oxygen_level"],labels=data["patient_name"],autopct="%1.2f%%")
plt.title("oxygen_level distribution")
plt.show()
```

```python
plt.boxplot(data["age"])
plt.xlabel("age")
plt.title("age distribution")
plt.show()
```

# From Kaggle Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("/content/patients.csv")
print(df)

print(df.head())

df["Recommendation"] = df.apply(
    lambda row:
    "Senior Care Advice" if row["age"] > 60 else
    "Improve Patient Experience" if row["satisfaction"] < 70 else
    "Standard Care Plan",
    axis=1
)

# Display recommendations
print("\nPatient Recommendations:\n")
print(df[["age", "satisfaction", "Recommendation"]])

# Count recommendations
recommendation_count = df["Recommendation"].value_counts()

print("\nRecommendation Counts:\n")
print(recommendation_count)
```

```python
recommendation_counts = df['Recommendation'].value_counts()
plt.pie(recommendation_counts.values, labels=recommendation_counts.index, autopct="%1.2f%%")
plt.title("Recommendation Distribution")
plt.show()
```

```python
plt.hist(df["satisfaction"], bins=10)
plt.xlabel("Satisfaction")
plt.ylabel("Frequency")
plt.title("Satisfaction Distribution")
plt.show()
```

```python
plt.scatter(df["age"], df["satisfaction"])
plt.xlabel("Age")
plt.ylabel("Satisfaction")
plt.title("Age vs Satisfaction")
plt.show()
```

```python
plt.boxplot(df["satisfaction"])
plt.ylabel("Satisfaction")
plt.title("Satisfaction Boxplot")
plt.show()
```

```python
plt.bar(df["age"], df["satisfaction"])
plt.xlabel("Age")
plt.ylabel("Satisfaction")
plt.title("Satisfaction by Age")
plt.show()
```
