```python
import statistics
import matplotlib.pyplot as plt
mem_names=["Sunil","Vardhan","Shanmukha","Shivaram","Vishnu"]
age=[30,10,21,22,19]
work_hours=[2,1.5,2.5,3,1]
cal_burn=[500,350,600,540,230]

#For Displaying Dataset
print("---FITNESS CENTER DATASET---")

for i in range(len(mem_names)):
    print(mem_names[i],age[i],work_hours[i],cal_burn[i])
#Mean Workout Hours
avg_work=sum(work_hours)/len(work_hours)
print("Average Workout Hours:",avg_work)

#Maximum Calories
max_cal=max(cal_burn)
print("Maximum Calories Burned:",max_cal)

#Minimum Calories
min_cal=min(cal_burn)
print("Minimum Calories Burned:",min_cal)

#Median Calories
med_cal=statistics.median(cal_burn)
print("Median Calories Burned:",med_cal)

#Standard Deviation
std_dev=statistics.stdev(cal_burn)
print("Standard Deviation of Calories Burned:",std_dev)

#Most Active Member
index=cal_burn.index(max_cal)
print("Most Active Member:",mem_names[index])

#bar chart
plt.bar(mem_names,cal_burn)
plt.xlabel("Members")
plt.ylabel("Calories Burned")
plt.title("Calories Burned by Members")
plt.show()

#line chart
plt.plot(mem_names,cal_burn)
plt.xlabel("Members")
plt.ylabel("Calories Burned")
plt.title("Calories Burned by Members")
plt.show()

#Scatter Plot
plt.scatter(mem_names,cal_burn)
plt.xlabel("Members")
plt.ylabel("Calories Burned")
plt.title("Calories Burned by Members")
plt.show()

```

```python
#Remove duplicates
df.drop_duplicates(inplace=True)

#
print(df.isnull())
print(df.isnull().sum())
```

```python
# Fill missing values
df = df.fillna(0)

# Convert datatype
df["reading score"] = df["reading score"].astype(int)

# Save cleaned data
df.to_csv("cleaned_data.csv", index=False)
print(df)
print("Data cleaned successfully")
```

```python
#Bar Chart
plt.figure(figsize=(5, 6))
plt.bar(df['race/ethnicity'], df['writing score'],color="red")
plt.xlabel('race/ethnicity')
plt.ylabel('writing Score')
plt.title('Writing Score by race/ethnicity')
plt.show()

#Line Chart
plt.plot(df['gender'], df['math score'])
plt.xlabel('Gender')
plt.ylabel('Math Score')
plt.title('Math Score by Gender')
plt.show()

#Scatter Plot
plt.scatter(df['writing score'], df['math score'])
plt.xlabel('writing score')
plt.ylabel('Math Score')
plt.title('Math Score by writing score')
plt.show()

#Histogram
plt.hist(df['math score'])
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.title('Distribution of Math Scores')
plt.show()

#Box Plot
plt.boxplot(df['math score'])
plt.ylabel('Math Score')
plt.title('Box Plot of Math Scores')
plt.show()

#Pie Chart
labels = df['gender'].value_counts().index
sizes = df['gender'].value_counts().values
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title('Gender Distribution')
plt.show()
```
