# Predictive Analysis

# User Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.DataFrame({
    "water_cons":[10,15,20,25,18,22],
    "Days":["Mon","Tues","Wednes","Thurs","Fri","Sat"]
})
print(df)
avg_water_cons=df["water_cons"].mean()
print("avg_water_cons=",avg_water_cons)
Total_water_uages=df["water_cons"].sum()
print("Total_water_uages=",Total_water_uages)
water_usage_increased=df["water_cons"]>avg_water_cons
df["water_usage_increased"]=water_usage_increased
sum_daily_water_increased=df.groupby("water_usage_increased")["water_cons"].sum()
print("sum_daily_water_increased=",sum_daily_water_increased)
print(df)
```

```python
#line chart
plt.plot(df["Days"],df["water_cons"])
plt.title("Daily Water Consumption")
plt.xlabel("Days")
plt.ylabel("Water Consumption")
plt.show()
```

```python
plt.pie(df["water_cons"],labels=df["Days"],autopct="%1.1f%%")
plt.title("Daily Water Consumption")
plt.show()
```

```python
plt.scatter(df["Days"],df["water_cons"])
plt.title("Daily Water Consumption")
plt.xlabel("Days")
plt.ylabel("Water Consumption")
plt.show()
```

```python
plt.boxplot(df["water_cons"])
plt.title("Daily Water Consumption")
plt.ylabel("Water Consumption")
plt.show()
```

```python
plt.bar(df["Days"],df["water_cons"])
plt.title("Daily Water Consumption")
plt.xlabel("Days")
plt.ylabel("Water Consumption")
plt.show()
```

```python
plt.hist(df["water_cons"])
plt.title("Daily Water Consumption")
plt.xlabel("Water Consumption")
plt.ylabel("Frequency")
plt.show()
```

# From Kaggle Dataset

```python
import pandas as pd

df = pd.read_csv("/content/House Price India.csv")

print(df.columns)

avg_house_price = df["Price"].mean()
print("avg_house_price =", avg_house_price)

total_house_price = df["Price"].sum()
print("total_house_price =", total_house_price)

price_above_average = df["Price"] > avg_house_price

df["price_above_average"] = price_above_average

sum_price_above_average = df.groupby("price_above_average")["Price"].sum()

print("sum_price_above_average =")
print(sum_price_above_average)

print(df)
```

```python
plt.plot(df["number of bedrooms"], df["Price"])
plt.title("House Price in India")
plt.xlabel("number of bedrooms")
plt.ylabel("Price")
plt.show()
```

```python
bedroom_counts = df['number of bedrooms'].value_counts()
plt.pie(bedroom_counts.values, labels=bedroom_counts.index, autopct="%1.2f%%")
plt.title("Number of Bedrooms Distribution")
plt.show()
```

```python
plt.scatter(df["number of bedrooms"], df["Price"])
plt.title("House Price in India")
plt.xlabel("number of bedrooms")
plt.ylabel("Price")
plt.show()
```

```python
plt.boxplot(df["Area of the house(excluding basement)"])
plt.title("Area of the house(excluding basement) in India")
plt.ylabel("Area of the house(excluding basement)")
plt.show()
```
