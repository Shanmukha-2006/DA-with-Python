```python
#swiggy/zomato Food Delivery Analysis
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df=pd.DataFrame({
    "Order_time":["9:00","9:10","12:00","1:00","5:00"],
    "Delivery_time":["9:45","9:55","12:20","1:40","5:20"],
    "Delivery_partner":["Swiggy","Swiggy","Zomato","Swiggy","Zomato"],
    "Cust_Ratings":[3.5,4.0,4.7,4.2,4.8]
})

df["Order_time"]=pd.to_datetime(df["Order_time"],format='%H:%M')
df["Delivery_time"]=pd.to_datetime(df["Delivery_time"],format='%H:%M')
df["Delivery_duration"]=df["Delivery_time"]-df["Order_time"]
print(df)

# Average delivery duration
avg_delivery_duration = df.groupby("Delivery_partner")["Delivery_duration"].mean()
print("\nAverage Delivery Duration by Partner:",avg_delivery_duration)

# Most delayed delivery partner
most_delayed_partner = avg_delivery_duration.idxmax()
print("\nMost Delayed Delivery Partner:", most_delayed_partner)

# Top rated delivery partner
top_rated_partner = df.groupby("Delivery_partner")["Cust_Ratings"].mean().idxmax()
print("\nTop Rated Delivery Partner:", top_rated_partner)

# Peak order hour(s)
peak_hours = df["Order_time"].dt.hour.mode().tolist()
print("\nPeak Order Hours:", peak_hours)
```

```python
plt.bar(df["Delivery_partner"],df["Cust_Ratings"])
plt.xlabel("Delivery Partner")
plt.ylabel("Customer Ratings")
plt.title("Customer Ratings for Different Delivery Partners")
plt.show()
```

```python
plt.scatter(df["Order_time"],df["Delivery_time"])
plt.xlabel("Order Time")
plt.ylabel("Delivery Time")
plt.title("Order Time vs Delivery Time")
plt.show()
```

```python
plt.plot(df["Delivery_partner"],df["Cust_Ratings"])
plt.xlabel("Delivery Partner")
plt.ylabel("Customer Ratings")
plt.title("Customer Ratings for Different Delivery Partners")
plt.show()
```

```python
plt.hist(df["Cust_Ratings"],bins=5)
plt.xlabel("Customer Ratings")
plt.ylabel("Frequency")
plt.title("Distribution of Customer Ratings")
plt.show()
```

```python
plt.boxplot(df["Cust_Ratings"])
plt.ylabel("Customer Ratings")
plt.title("Boxplot of Customer Ratings")
plt.show()
```

```python
plt.pie(df["Cust_Ratings"],labels=df["Delivery_partner"],autopct="%1.1f%%")
plt.title("Customer Ratings for Different Delivery Partners")
plt.show()
```

```python
#Online Shopping Behaviour Analysis
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df=pd.DataFrame({
    "prod_purchase":["Mobile","Shoes","Mobile","Laptop","Watch"],
    "card_abandoment":[20,35,40,100,5],
    "cust_ratings":[4.5,4.0,4.4,4.6,3.5],
    "pay_method":["UPI","Cash","UPI","UPI","Cash"],
})
print(df)

#Most sold Products
most_sold_products=df["prod_purchase"].mode().values[0]
print("\nmost_sold_products:",most_sold_products)

#High revenue customers
high_revenue_customers=df.groupby("pay_method")["card_abandoment"].sum().idxmax()
print("\nhigh_revenue_customers:",high_revenue_customers)

#most used payment method
most_used_payment_method=df["pay_method"].mode().values[0]
print("\nmost_used_payment_method:",most_used_payment_method)

#Produts with poor ratings
products_with_poor_ratings=df[df["cust_ratings"]<4.0]["prod_purchase"].values[0]
print("\nproducts_with_poor_ratings:",products_with_poor_ratings)

```

```python
#bar plot
plt.bar(df["prod_purchase"],df["card_abandoment"])
plt.xlabel("Products")
plt.ylabel("Card Abandoment")
plt.title("Card Abandoment for Different Products")
plt.show()
```

```python
#line chart
plt.plot(df["prod_purchase"],df["cust_ratings"])
plt.xlabel("Products")
plt.ylabel("Customer Ratings")
plt.title("Customer Ratings for Different Products")
plt.show()
```

```python
#box plot
plt.boxplot(df["cust_ratings"])
plt.ylabel("Customer Ratings")
plt.title("Boxplot of Customer Ratings")
plt.show()
```

```python
#histogram
plt.hist(df["cust_ratings"],bins=5)
plt.xlabel("Customer Ratings")
plt.ylabel("Frequency")
plt.title("Distribution of Customer Ratings")
plt.show()
```

```python
#pie chart
plt.pie(df["cust_ratings"],labels=df["prod_purchase"],autopct="%1.1f%%")
plt.title("Customer Ratings for Different Products")
plt.show()
```

```python
#Bank Transaction Analysis
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df=pd.DataFrame({
    "Trans_Amount":[100,6000,25,500,5000],
    "Location":["Hyderabad","Andhra Pradesh","Andhra Pradesh","Hyderabad","Tamilnadu"],
    "Time":["9:00","11:00","13:00","19:00","22:00"],
    "Account_Act":["InActive","Active","Active","InActive","Active"],
})
print(df)
#Unusual transactions
Unusual_trans=df.loc[df["Trans_Amount"]>5000,"Location"].values[0]
print("\nUnusual_trans:",Unusual_trans)

#High-risk acounts
high_risk_accounts=df.loc[df["Account_Act"]=="InActive","Location"].values[0]
print("\nhigh_risk_accounts:",high_risk_accounts)

#fraud patterns
fraud_patterns=df.loc[df["Trans_Amount"]<=50,"Location"].values[0]
print("\nfraud_patterns:",fraud_patterns)
```

```python
# Supermarket Inventory & Sales Analysis

import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Product": ["Rice", "Milk", "Bread", "Oil", "Biscuits"],
    "Daily_Sales": [120, 50, 200, 180, 300],
    "Product_Stock": [500, 80, 250, 400, 50],
    "Expiry_Date": ["2026-12-16", "2027-01-15", "2026-10-14", "2027-11-30", "2026-06-24"],
    "Seasonal_Demand": ["Medium", "Low", "High", "Medium", "High"]
})

print(data)

avg_sales = data["Daily_Sales"].mean()

fast_moving = data[data["Daily_Sales"] > avg_sales]
print("\nFast Moving Products:")
print(fast_moving[["Product", "Daily_Sales"]])

slow_moving = data[data["Daily_Sales"] <= avg_sales]
print("\nSlow Moving Products:")
print(slow_moving[["Product", "Daily_Sales"]])


data["Days_Left"] = data["Product_Stock"] / data["Daily_Sales"]

print("\nStock Remaining (Days):")
print(data[["Product", "Days_Left"]])

out_of_stock = data[data["Days_Left"] < 5]

print("\nProducts Likely to Go Out of Stock Soon:")
print(out_of_stock[["Product", "Days_Left"]])
```

```python
plt.figure(figsize=(8,5))
plt.bar(data["Product"], data["Daily_Sales"])
plt.title("Daily Sales by Product")
plt.xlabel("Product")
plt.ylabel("Daily Sales")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(data["Product"], data["Daily_Sales"])
plt.title("Daily Sales by Product")
plt.xlabel("Product")
plt.ylabel("Daily Sales")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.pie(data["Daily_Sales"], labels=data["Product"], autopct="%1.1f%%")
plt.title("Daily Sales by Product")
plt.show()
```

```python
plt.hist(data["Product_Stock"], bins=5)
plt.title("Distribution of Stock")
plt.xlabel("Daily Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.boxplot(data["Daily_Sales"])
plt.title("Boxplot of Daily Sales")
plt.ylabel("Daily Sales")
plt.show()
```
